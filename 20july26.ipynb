```python
my_dict = {}
for i in range(1, 6):
    multiplies = {i: i * j for j in range(1, 6)}
    my_dict[i] = multiplies
    print(my_dict)    
```

```python
dict_5 = {x: {x*j for j in range(1, 6)} for x in range(1, 6)}
print(dict_5)
```

```python
dict_6 = {x: tuple({x**j for j in range(1, 6)}) for x in range(1, 6)}
print(dict_6)
```

```python
dict_7 = {x: x**2 for x in range(1, 11) }
even_numbers = {key: value for key, value in dict_7.items() if key % 2 == 0}
print(even_numbers)
```

```python
dict_8 = {i: i**2 for i in range(1, 6)}
swapped = {value: key for key, value in dict_8.items()}
print(swapped)
```

```python
from collections import defaultdict
d = defaultdict(list)
d["Fruits"].append("Apple")
d["Fruits"].append("Mango")
d["Vegetables"].append("Carrot")
d["Vegetables"].append("Potato")
print(d)
```

```python
def count_characters(string):
    count = {}

    for char in string:
        if char in count:
            count[char] += 1
        else:
            count[char] = 1

    return count
result = count_characters('hello')

print(result)
```
