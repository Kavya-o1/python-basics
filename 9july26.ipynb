```python
marks = [95,82,-1,67,45,78,91,-1,38,100]
pl = []
ps = 0
for student in marks:
    if student == -1:
        continue
    pl.append(student)
for student in pl:
    if student == 100:
        break
    print(student)
    
    if (90<student<100):
        print("A Grade")
        ps += 1
    elif (80<student<90):    
        print("B Grade")
        ps += 1
    elif (70<student<80):
        print("C Grade")
        ps += 1
    else:
        print("D Grade")
        ps += 1
print("Total Students Passed:", ps)

avg = sum(pl)/len(pl)
print("Average Marks of students:", avg)
```
